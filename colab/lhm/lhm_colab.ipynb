# 写真（1〜8枚）か動画から、全身の立体人物を作る（LHM++・試作）

1. 「ランタイム → ランタイムのタイプを変更」で **L4** を選ぶ
2. 「1. 準備」を実行（初回15〜30分。同じセッションなら2回目からはすぐ終わる）
3. 「2. 立体人物を作る」を実行して、**1人分**の写真か動画を選ぶ
   - 写真：**同じ日・同じ服・同じ髪型**のものを1〜8枚。正面・斜め・横・後ろがあるほど良い。1枚目には頭から足まで写った写真が自動で選ばれる
   - 動画：その人が動いたり振り向いたりしている動画1本（切り出しツールのクリップでOK）。くっきりしたコマを8枚自動で選ぶ
   - 何人か写っていても、いちばん大きく写っている人だけを使う
4. ぐるっと一周回る確認動画が出て、立体人物（〜_lhm.ply）がダウンロードされる。人ごとに「2.」をくり返す

写真・動画はこのColabの中だけで処理します（Googleドライブにはつなぎません）。使い終わったら「ランタイム → ランタイムを接続解除して削除」。
