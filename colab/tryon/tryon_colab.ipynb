# 着せ替えツール
ランタイムは **L4・ハイメモリ** にしてください。写真はこのColabの中だけで処理します。
- qwen（おすすめ）: 日本の家族写真・複数人に強い。服の写真を渡して着せることもできる
- kontext: 一か所ずつ少しずつ直すのが得意（Hugging Faceで利用規約への同意とトークンが必要）

控えめな指示ほどうまくいきます。姿勢を変える・一人だけ替える・細かい柄は苦手です。

```
# 1. 準備
!pip -q install -U git+https://github.com/huggingface/diffusers transformers accelerate peft sentencepiece
!rm -rf /content/ppr && git clone -q --depth 1 https://github.com/noshukukangen-lgtm/pop-photo-room /content/ppr
import sys; sys.path.insert(0, '/content/ppr/colab/tryon')
import tryon_tool as T
print('指示の例:', list(T.EXAMPLES))
```

```
# 1b. Kontextを使うときだけ: Hugging Faceのトークンを入れる
from huggingface_hub import notebook_login; notebook_login()
```

```
# 2. 写真をアップロード（服の写真を使うなら一緒に）
from google.colab import files
up = files.upload()
names = list(up); print(names)
```

```
# 3. 着せ替え（指示は EXAMPLES の名前か、英語の文章）
PHOTO = names[0]
PROMPT = '上下ぴったり'
T.edit(PHOTO, PROMPT, model='qwen', seed=0)
# 服の写真を着せる例: T.edit(PHOTO, '服の写真を着せる', model='qwen', clothes=names[1])
# 結果が気に入らなければ seed を 1, 2 … と変えて再実行
```

```
# 4. AIどうしを比べる（Kontextはトークン入力後）
T.compare(PHOTO, PROMPT, models=('qwen','kontext'))
```
