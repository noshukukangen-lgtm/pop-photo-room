# 本物の動画を、全部のコマで立体にする

1. 「ランタイム → ランタイムのタイプを変更」で **L4** を選ぶ
2. 下のセルを実行して、動画（切り出しツールで作ったクリップなど）を選ぶ
3. 「長いまま」にチェックを入れて「ファイル名の時間で作る」→ 音声つき・全部のコマで立体の「〜_3d.mp4」
   チェックを外すと、つなぎ目がきれいな範囲でできるだけ長くループする（2〜30秒）「〜_loop_3d.mp4」
4. できた動画をアプリの「動画（ループ・VR180）」で読み込む

VR180の動画は、これまでどおりの作り方になります。Googleドライブには一切つなぎません。
