# LoRA学習ツール（試作）
ランタイムは **L4**。写真はこのColabの中だけで処理します（ドライブにはつなぎません）。

- **特定の人**（mode='person'）：NASで分けたその人の写真をZIPで。20〜40枚が目安。子どもは1年分など期間を区切ると顔がぼやけない

学習はL4で1〜2時間ほど。使い終わったら「ランタイムを接続解除して削除」。
